## **Line Plot**

A line plot is created using the `plot()` function.

The `show()` function is used to display the graph.

```python
# AP24110010105

from matplotlib import pyplot as plt

x = [5, 2, 9, 4, 7]
y = [10, 5, 8, 4, 2]

plt.plot(x, y)

plt.savefig("line_plot.png")

plt.show()
```

## **Line Plot with Marker, Labels and Grid**

A line plot can be customized using:
- Markers
- X-axis labels
- Y-axis labels
- Gridlines
- Title

```python
# AP24110010105

import matplotlib.pyplot as plt

X = [12, 34, 23, 45, 67, 89]
Y = [1, 3, 67, 78, 7, 5]

plt.plot(X, Y, marker='o', markerfacecolor='r')

plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")

plt.grid(linestyle='--', linewidth=0.5)

plt.title("Number of Users of a Particular Language")

plt.show()
```

## **Plotting Multiple Lines in a Line Plot**

Multiple lines can be plotted in the same graph using multiple `plot()` functions.

Different line styles can be used:
- `-` Solid line
- `--` Dashed line
- `-.` Dash-dot line
- `:` Dotted line

```python
# AP24110010105

import matplotlib.pyplot as plt
import numpy as np

x = [1, 2, 3, 4, 5]
y = [3, 3, 3, 3, 3]

plt.plot(x, y, label="line 1", linestyle="-")
plt.plot(y, x, label="line 2", linestyle="--")
plt.plot(x, np.sin(x), label="curve 1", linestyle="-.")
plt.plot(x, np.cos(x), label="curve 2", linestyle=":")

plt.legend()

plt.show()
```

## **Bar Plot**

## **Horizontal Bar Chart**

A horizontal bar chart is created using the `barh()` function.

```python
# AP24110010105

import matplotlib.pyplot as plt

y = ['one', 'two', 'three', 'four', 'five']
x = [5, 24, 35, 67, 12]

plt.barh(y, x)

plt.ylabel("Pen Sold")
plt.xlabel("Price")
plt.title("Horizontal Bar Graph")

plt.show()
```

## **Stacked Bar Chart**

A stacked bar chart displays multiple categories stacked on top of each other.

```python
# AP24110010105

import matplotlib.pyplot as plt
import pandas as pd

data = [
    ['A', 10, 20, 10, 26],
    ['B', 20, 25, 15, 21],
    ['C', 12, 15, 19, 6],
    ['D', 10, 18, 11, 19]
]

df = pd.DataFrame(
    data,
    columns=['Team', 'Round 1', 'Round 2', 'Round 3', 'Round 4']
)

print(df)

df.plot(
    x='Team',
    kind='bar',
    stacked=True,
    title='Stacked Bar Graph by DataFrame'
)

plt.show()
```

## **Two Bar Plots in a Graph**

Two different sets of bars can be plotted in the same graph to compare two categories.

```python
# AP24110010105

from matplotlib import pyplot as plt

plt.bar(
    [0.25, 1.25, 2.25, 3.25, 4.25],
    [50, 40, 70, 80, 20],
    label="BMW",
    width=0.5
)

plt.bar(
    [0.75, 1.75, 2.75, 3.75, 4.75],
    [80, 20, 20, 50, 60],
    label="Audi",
    width=0.5
)

plt.legend()

plt.xlabel("Days")
plt.ylabel("Distance (kms)")
plt.title("Information")

plt.show()
```

## **Histogram**

A histogram is used for the graphical representation of the frequency distribution of data.

The `hist()` function is used to create a histogram.

```python
# AP24110010105

from matplotlib import pyplot as plt

y = [10, 5, 5, 8, 4, 10, 10, 2]

plt.hist(y)

plt.show()
```

## **Plotting Two Histograms in the Same Graph**

Two histograms can be plotted in the same graph to compare two groups of data.

```python
# AP24110010105

import matplotlib.pyplot as plt

age_g1 = [
    1, 3, 5, 10, 15, 17, 18, 16, 19, 21,
    23, 28, 30, 31, 33, 38, 32, 40, 45, 43,
    49, 55, 53, 63, 66, 85, 80, 57, 75, 93, 95
]

age_g2 = [
    6, 4, 15, 17, 19, 21, 28, 23, 31, 36,
    39, 32, 50, 56, 59, 74, 79, 34, 98, 97,
    95, 67, 69, 92, 45, 55, 77, 76, 85
]

plt.hist(age_g1, label='Age Group 1', bins=14)
plt.hist(age_g2, label='Age Group 2', bins=14)

plt.legend()

plt.show()
```

## **Scatter Plot**

A scatter plot displays individual data points using X and Y coordinates.

The `scatter()` function is used to create a scatter plot.

```python
# AP24110010105

from matplotlib import pyplot as plt

x = [5, 2, 9, 4, 7]
y = [10, 5, 8, 4, 2]

plt.scatter(x, y)

plt.show()
```

## **Scatter Plot with Two Groups**

Two different groups can be represented using two scatter plots.

```python
# AP24110010105

import matplotlib.pyplot as plt

x = [1, 1.5, 2, 2.5, 3, 3.5, 3.6]
y = [7.5, 8, 8.5, 9, 9.5, 10, 10.5]

x1 = [8, 8.5, 9, 9.5, 10, 10.5, 11]
y1 = [3, 3.5, 3.7, 4, 4.5, 5, 5.2]

plt.scatter(x, y, label='High Income Low Saving')
plt.scatter(x1, y1, label='Low Income High Savings')

plt.xlabel('Saving * 100')
plt.ylabel('Income * 1000')
plt.title('Scatter Plot')

plt.legend()

plt.show()
```

## **Pie Plot**

A pie plot represents data as slices of a circle.

The `pie()` function is used to create a pie plot.

```python
# AP24110010105

import matplotlib.pyplot as plt

slices = [7, 2, 2, 13]

activities = [
    'Sleeping',
    'Eating',
    'Working',
    'Playing'
]

plt.pie(
    slices,
    labels=activities
)

plt.title('Pie Plot')

plt.show()
```

## **Heatmap**

A heatmap is a graphical representation of data using colors to visualize the values of a matrix.

Different colors represent different values in the data.

```python
# AP24110010105

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = np.random.randint(
    low=1,
    high=100,
    size=(10, 10)
)

print("The data to be plotted:\n")
print(data)

hm = sns.heatmap(
    data=data,
    annot=True
)

plt.show()
```

## **Important Heatmap Parameters**

Important parameters of `seaborn.heatmap()` include:

- `data` - 2D dataset
- `vmin` - Minimum value for the colormap
- `vmax` - Maximum value for the colormap
- `cmap` - Mapping from data values to colors
- `center` - Center value of the colormap
- `annot` - Displays data values inside cells
- `fmt` - Formatting code for annotations
- `linewidths` - Width of lines between cells
- `linecolor` - Color of lines between cells
- `cbar` - Controls whether the colorbar is displayed

## **DataFrame from Dictionary of Series**

```python
# AP24110010105

info = {
    'one': pd.Series(
        [10, 20, 30, 40, 50, 60],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    ),

    'two': pd.Series(
        [10, 20, 30, 40, 50, 60, 70, 80],
        index=['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']
    )
}

d1 = pd.DataFrame(info)

print(d1)
```

## **Column Selection**

```python
# AP24110010105

print("Column one:")
print(d1['one'])

print("\nColumn two:")
print(d1['two'])
```

## **Column Addition**

```python
# AP24110010105

info = {
    'one': pd.Series(
        [10, 20, 30, 40, 50],
        index=['a', 'b', 'c', 'd', 'e']
    ),

    'two': pd.Series(
        [10, 20, 30, 40, 50, 60],
        index=['a', 'b', 'c', 'd', 'e', 'f']
    )
}

df = pd.DataFrame(info)

print("Original DataFrame:")
print(df)

df['three'] = pd.Series(
    [25, 50, 75],
    index=['a', 'b', 'c']
)

print("\nAfter adding third column:")
print(df)

df['four'] = df['one'] + df['three']

print("\nAfter adding fourth column:")
print(df)
```

## **Addition of Rows**

```python
# AP24110010105

d = pd.DataFrame({
    'A': [10, 20, 30],
    'B': [40, 50, 60]
})

d2 = pd.DataFrame({
    'A': [70, 80, 90],
    'B': [100, 110, 120]
})

d = pd.concat([d, d2], ignore_index=True)

print(d)
```

## **Deletion of Rows**

```python
# AP24110010105

a_info = pd.DataFrame(
    [[10, 20], [30, 40]],
    columns=['x', 'y']
)

b_info = pd.DataFrame(
    [[50, 60], [70, 80]],
    columns=['x', 'y']
)

a_info = pd.concat([a_info, b_info], ignore_index=True)

print("Original DataFrame:")
print(a_info)

a_info = a_info.drop(0)

print("\nAfter deleting row 0:")
print(a_info)
```

## **Working with Columns**

```python
# AP24110010105

data = {
    'Gender': ['M', 'F', 'M'],
    'Emp_ID': ['E11', 'E12', 'E13'],
    'Age': [22, 24, 23]
}

df = pd.DataFrame(
    data,
    columns=['Emp_ID', 'Gender', 'Age']
)

print(df)

print("\nEmp_ID column:")
print(df.Emp_ID)

print("\nFirst Emp_ID:")
print(df.Emp_ID[0])
```

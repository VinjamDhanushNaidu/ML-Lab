## **Image Display**

```python
#AP24110010105

import numpy as np
import matplotlib.pyplot as plt

# Create a dark 300 x 300 RGB image
image = np.zeros((300, 300, 3))

# Create coordinate values
y, x = np.ogrid[:300, :300]

# Create a large moon
moon = (x - 220) ** 2 + (y - 70) ** 2 < 40 ** 2
image[moon] = [1, 0.9, 0.5]

# Create the first mountain
mountain1 = y > (250 - np.abs(x - 100) * 1.4)
image[mountain1] = [0.15, 0.2, 0.3]

# Create the second mountain
mountain2 = y > (230 - np.abs(x - 210) * 1.1)
image[mountain2] = [0.25, 0.3, 0.4]

# Add snow on the mountain peaks
snow1 = (y > (250 - np.abs(x - 100) * 1.4)) & (y < (190 + np.abs(x - 100) * 0.6))
image[snow1] = [0.8, 0.85, 0.9]

# Add small stars
stars = (
    ((x - 50) ** 2 + (y - 50) ** 2 < 4) |
    ((x - 120) ** 2 + (y - 35) ** 2 < 4) |
    ((x - 175) ** 2 + (y - 80) ** 2 < 4) |
    ((x - 270) ** 2 + (y - 120) ** 2 < 4) |
    ((x - 80) ** 2 + (y - 100) ** 2 < 4)
)

image[stars] = [1, 1, 1]

# Display the artwork
plt.imshow(image)

# Hide the axes
plt.axis("off")

# Show the image
plt.show()
```

## **Image Cropping**

```python
#AP24110010105
import matplotlib.pyplot as plt

# Use the image created earlier
img_array = image

# Get image dimensions
height, width = img_array.shape[:2]

# Find the starting point for the center 200x200 crop
start_y = (height - 200) // 2
start_x = (width - 200) // 2

# Crop the center 200x200 region
crop = img_array[start_y:start_y + 200, start_x:start_x + 200]

# Display the crop
plt.imshow(crop)
plt.axis("off")
plt.show()
```

##**4. Comparing Two Sentences**

```python
#AP24110010105
import numpy as np

# Similar pair
sentence1 = "I enjoy learning Python programming"
sentence2 = "I like learning Python coding"

# Very different pair
sentence3 = "I enjoy playing football"
sentence4 = "The computer needs more memory"

def cosine_similarity(s1, s2):
    words1 = s1.lower().split()
    words2 = s2.lower().split()

    vocab = sorted(set(words1 + words2))

    vec1 = np.array([words1.count(word) for word in vocab])
    vec2 = np.array([words2.count(word) for word in vocab])

    return np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))

# Calculate similarity
similarity1 = cosine_similarity(sentence1, sentence2)
similarity2 = cosine_similarity(sentence3, sentence4)

print("Similar sentences:", similarity1)
print("Different sentences:", similarity2)
```

##**5. Most Frequent words**

```python
#AP24110010105
import numpy as np

# Paragraph
text = """
Machine learning helps computers learn from data.
Machine learning is useful in many real-world applications.
Data helps machine learning models make better predictions.
Learning from data is an important part of machine learning.
"""

# 1. Tokenize the paragraph
tokens = text.lower().split()

# 2. Build vocabulary and word-count vector
vocabulary = np.array(sorted(set(tokens)))
word_counts = np.array([tokens.count(word) for word in vocabulary])

# 3. Find the top 3 most frequent words
top_indices = np.argsort(word_counts)[::-1][:3]

print("Vocabulary:", vocabulary)
print("Word counts:", word_counts)

print("\nTop 3 most frequent words:")
for i in top_indices:
    print(vocabulary[i], ":", word_counts[i])
```

# **Mini Project : Simple Text Similarity Checker**

```python
#AP24110010105
import numpy as np

# Three sentences of our choice
sentence1 = "I enjoy learning Python programming"
sentence2 = "I enjoy learning Python coding"
sentence3 = "I like playing football with friends"

# Tokenize the sentences
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

# Create one shared vocabulary
vocabulary = np.array(sorted(set(tokens1 + tokens2 + tokens3)))

# Create bag-of-words vectors
vector1 = np.array([tokens1.count(word) for word in vocabulary])
vector2 = np.array([tokens2.count(word) for word in vocabulary])
vector3 = np.array([tokens3.count(word) for word in vocabulary])

# Function to calculate cosine similarity
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Calculate similarity for all three pairs
sim12 = cosine_similarity(vector1, vector2)
sim13 = cosine_similarity(vector1, vector3)
sim23 = cosine_similarity(vector2, vector3)

# Display results
print("Sentence 1:", sentence1)
print("Sentence 2:", sentence2)
print("Sentence 3:", sentence3)

print("\nCosine Similarities:")
print("Sentence 1 & 2:", sim12)
print("Sentence 1 & 3:", sim13)
print("Sentence 2 & 3:", sim23)

# Find the most similar pair
scores = [sim12, sim13, sim23]
pairs = ["Sentence 1 and Sentence 2",
         "Sentence 1 and Sentence 3",
         "Sentence 2 and Sentence 3"]

best_pair = np.argmax(scores)

print("\nMost similar pair:", pairs[best_pair])
print("Similarity score:", scores[best_pair])
```

##**1.Why do we need to convert text into numbers before a machine learning model can use it?**

Machine learning models work with numbers, not raw text. Converting text into numbers lets the model process the words and find patterns or similarities in the data.


**2.What information does a bag-of-words vector lose about the original sentence? (Hint: think about what `split()` and counting alone doesn't capture.)**

A bag-of-words vector loses the order and context of the words. It only keeps how many times each word appears.

For example, “Dog bites man” and “Man bites dog” would have the same bag-of-words counts, even though their meanings are different.


**3.Why might cosine similarity be a better similarity measure than simply comparing raw word-count differences?**

Cosine similarity is better because it compares the direction of the word vectors, not just the difference in their counts. So, it can recognize similar text even when the sentences have different lengths.
